## Correlations

```python
# ============================================================
# 1. Select variables
# ============================================================

# AOD variables from MERRA-2
aod_var_list = ["TOTEXTTAU", "DUEXTTAU", "BCEXTTAU", 'OCEXTTAU', 'SSEXTTAU', 'SUEXTTAU']

# Soil water variable
print("Soil water variables:", list(soil_water.data_vars))

sw_var = list(soil_water.data_vars)[0]
print("Using soil water variable:", sw_var)

sw_max = soil_water[sw_var]
```

```python
# ============================================================
# 2. Helper functions
# ============================================================

def rename_coords_if_needed(da):
    """
    Rename coordinates to standard names: time, lat, lon
    """
    rename_dict = {}

    if "valid_time" in da.dims or "valid_time" in da.coords:
        rename_dict["valid_time"] = "time"

    if "longitude" in da.dims or "longitude" in da.coords:
        rename_dict["longitude"] = "lon"

    if "latitude" in da.dims or "latitude" in da.coords:
        rename_dict["latitude"] = "lat"

    if rename_dict:
        da = da.rename(rename_dict)

    return da


def convert_lon_to_180(da):
    """
    Convert longitude from 0-360 to -180 to 180 if needed
    """
    if float(da.lon.max()) > 180:
        da = da.assign_coords(lon=(((da.lon + 180) % 360) - 180))
        da = da.sortby("lon")
    return da


def lag1_autocorr(da):
    """
    Calculate lag-1 autocorrelation along time for each pixel
    """
    x0 = da.isel(time=slice(0, -1))
    x1 = da.isel(time=slice(1, None))

    # Reset time coordinate so paired values align correctly
    x0 = x0.assign_coords(time=np.arange(x0.sizes["time"]))
    x1 = x1.assign_coords(time=np.arange(x1.sizes["time"]))

    return xr.corr(x0, x1, dim="time")
```

```python
# ============================================================
# 3. Main computation function
# ============================================================

def compute_spearman_map_with_effN(aod_da, sw_da, extent):
    """
    Compute the Spearman correlation between AOD and soil water
    at every grid cell.
    """

    lon_min, lon_max, lat_min, lat_max = extent

    # Rename coordinates
    aod_da = rename_coords_if_needed(aod_da)
    sw_da = rename_coords_if_needed(sw_da)

    # Convert longitude if needed
    aod_da = convert_lon_to_180(aod_da)
    sw_da = convert_lon_to_180(sw_da)

    # Sort coordinates
    aod_da = aod_da.sortby("lat").sortby("lon")
    sw_da = sw_da.sortby("lat").sortby("lon")

    # Subset study domain
    aod_sub = aod_da.sel(
        lon=slice(lon_min, lon_max),
        lat=slice(lat_min, lat_max)
    )

    sw_sub = sw_da.sel(
        lon=slice(lon_min, lon_max),
        lat=slice(lat_min, lat_max)
    )

    print("AOD subset sizes:", aod_sub.sizes)
    print("SW subset sizes:", sw_sub.sizes)

    # Interpolate soil water to the AOD grid
    sw_on_aod_grid = sw_sub.interp(
        lon=aod_sub["lon"],
        lat=aod_sub["lat"]
    )

    print(
        "Interpolated SW sizes:",
        sw_on_aod_grid.sizes
    )

    # --------------------------------------------------------
    # Align daily timestamps
    # --------------------------------------------------------
    aod_sub["time"] = aod_sub["time"].dt.floor("D")

    sw_on_aod_grid["time"] = (
        sw_on_aod_grid["time"].dt.floor("D")
    )

    aod_aligned, sw_aligned = xr.align(
        aod_sub,
        sw_on_aod_grid,
        join="inner"
    )

    print("Common days:", len(aod_aligned["time"]))
    print("AOD aligned sizes:", aod_aligned.sizes)
    print("SW aligned sizes:", sw_aligned.sizes)

    print("Common time range:", str(aod_aligned["time"].min().values), "to",
        str(aod_aligned["time"].max().values) )

    # Spearman correlation map
    aod_rank = aod_aligned.rank(dim="time")
    sw_rank = sw_aligned.rank(dim="time")

    spearman_corr_map = xr.corr(aod_rank, sw_rank, dim="time")

    spearman_corr_map.name = "spearman_corr_map"

    spearman_min = float( spearman_corr_map.min(skipna=True).compute().item())

    spearman_max = float( spearman_corr_map.max(skipna=True).compute().item())

    print("Spearman min/max:", spearman_min,spearman_max)

    # Mean negative and positive correlations
    negative_correlations = spearman_corr_map.where(
        spearman_corr_map < 0 )

    positive_correlations = spearman_corr_map.where(
        spearman_corr_map > 0 )

    mean_negative_corr = float(
        negative_correlations
        .mean(dim=["lat", "lon"], skipna=True).compute().item())

    mean_positive_corr = float(
        positive_correlations
        .mean(dim=["lat", "lon"], skipna=True).compute().item())

    n_negative_pixels = int(
        (spearman_corr_map < 0)
        .sum(dim=["lat", "lon"]).compute().item())

    n_positive_pixels = int(
        (spearman_corr_map > 0)
        .sum(dim=["lat", "lon"]).compute().item())

    print("Mean negative Spearman correlation:",f"{mean_negative_corr:.3f}")

    print("Mean positive Spearman correlation:",f"{mean_positive_corr:.3f}")

    print("Number of negative pixels:",n_negative_pixels)

    print("Number of positive pixels:",n_positive_pixels)

    # Number of valid paired samples
    valid_pairs = (np.isfinite(aod_aligned)
        & np.isfinite(sw_aligned))

    n_valid = valid_pairs.sum(dim="time")

    # Lag-1 autocorrelation
    aod_lag1 = lag1_autocorr(aod_aligned)
    sw_lag1 = lag1_autocorr(sw_aligned)

    print("AOD lag-1 min/max:",
        float(aod_lag1.min(skipna=True)),
        float(aod_lag1.max(skipna=True)))

    print("SW lag-1 min/max:",
        float(sw_lag1.min(skipna=True)),
        float(sw_lag1.max(skipna=True)))

    # Effective sample size
    r1_product = aod_lag1 * sw_lag1

    n_eff = n_valid * ((1 - r1_product) / (1 + r1_product) )

    n_eff = n_eff.where(np.isfinite(n_eff))
    n_eff = n_eff.where(n_eff >= 3)
    n_eff = n_eff.where(n_eff <= n_valid)

    print("Raw N min/max:",int(n_valid.min(skipna=True)), int(n_valid.max(skipna=True)))

    print("Effective N min/max:",float(n_eff.min(skipna=True)), float(n_eff.max(skipna=True)))

    print("Mean effective N:",float(n_eff.mean(skipna=True)))

    # Autocorrelation-adjusted p-value
    rho = spearman_corr_map.clip(
        min=-0.999999,
        max=0.999999
    )

    df_eff = n_eff - 2

    t_stat_eff = rho * np.sqrt(
        df_eff / (1 - rho**2)
    )

    p_map_spearman_eff = xr.apply_ufunc(
        lambda t, df_: 2 * stats.t.sf(
            np.abs(t),
            df_
        ),
        t_stat_eff,
        df_eff,
        vectorize=True,
        dask="allowed",
        output_dtypes=[float]
    )

    p_map_spearman_eff = (p_map_spearman_eff.where(n_eff >= 3))

    p_map_spearman_eff.name = ("p_value_spearman_autocorr_adjusted")

    print("Adjusted Spearman p-value min/max:", float(p_map_spearman_eff.min(skipna=True)),
        float(p_map_spearman_eff.max(skipna=True)))

    # Return results
    return {
        "spearman_corr_map": spearman_corr_map,
        "p_map_spearman_eff": p_map_spearman_eff,
        "n_eff": n_eff,
        "n_valid": n_valid,
        "aod_aligned": aod_aligned,
        "sw_aligned": sw_aligned,
        "spearman_min": spearman_min,
        "spearman_max": spearman_max,
        "mean_negative_corr": mean_negative_corr,
        "mean_positive_corr": mean_positive_corr,
        "n_negative_pixels": n_negative_pixels,
        "n_positive_pixels": n_positive_pixels,
    }
```

```python
# ============================================================
# 4. Compute results for each AOD variable
# ============================================================

extent = [-123, -98, 25, 45]

results_dict = {}

for aod_var in aod_var_list:

    print("\n" + "=" * 70)
    print(f"Processing {aod_var}")
    print("=" * 70)

    aod_max = max_aod_merra[aod_var]

    results_dict[aod_var] = (
        compute_spearman_map_with_effN(
            aod_da=aod_max,
            sw_da=sw_max,
            extent=extent
        )
    )


# Summary table

correlation_summary_rows = []

for aod_var in aod_var_list:

    result = results_dict[aod_var]

    correlation_summary_rows.append({
        "AOD variable": aod_var,
        "Minimum correlation": result["spearman_min"],
        "Maximum correlation": result["spearman_max"],
        "Mean negative correlation": result["mean_negative_corr"],
        "Mean positive correlation": result["mean_positive_corr"],
        "Negative pixels": result["n_negative_pixels"],
        "Positive pixels": result["n_positive_pixels"],
    })


correlation_summary_df = pd.DataFrame(
    correlation_summary_rows
)

correlation_columns = [
    "Minimum correlation",
    "Maximum correlation",
    "Mean negative correlation",
    "Mean positive correlation",
]

correlation_summary_df[correlation_columns] = (
    correlation_summary_df[correlation_columns]
    .round(3)
)

correlation_summary_df
```

```python
# ============================================================
# Percentages of significant negative and positive correlations
# ============================================================

southwest_rows = []

for aod_var in aod_var_list:

    corr = results_dict[aod_var]["spearman_corr_map"]
    pval = results_dict[aod_var]["p_map_spearman_eff"]

    southwest = (
        (corr.lon >= lon_box_min)
        & (corr.lon <= lon_box_max)
        & (corr.lat >= lat_box_min)
        & (corr.lat <= lat_box_max)
    )

    valid = southwest & np.isfinite(corr) & np.isfinite(pval)

    significant_negative = valid & (pval < 0.05) & (corr < 0)
    significant_positive = valid & (pval < 0.05) & (corr > 0)

    n_valid = int(valid.sum().compute().item())
    n_negative = int(significant_negative.sum().compute().item())
    n_positive = int(significant_positive.sum().compute().item())

    southwest_rows.append({
        "AOD variable": aod_var,
        "Valid Southwest pixels": n_valid,
        "Significant negative pixels": n_negative,
        "Significant positive pixels": n_positive,
        "Significant negative pixels [%]": 100 * n_negative / n_valid,
        "Significant positive pixels [%]": 100 * n_positive / n_valid,
    })

southwest_significance_summary = pd.DataFrame(southwest_rows)

southwest_significance_summary[
    [
        "Significant negative pixels [%]",
        "Significant positive pixels [%]",
    ]
] = southwest_significance_summary[
    [
        "Significant negative pixels [%]",
        "Significant positive pixels [%]",
    ]
].round(1)

southwest_significance_summary
```

```python
# ============================================================
# Density distributions of AOD–soil water correlations
# Southwest box only
# ============================================================

# Southwest boundaries
lon_box_min, lon_box_max = -120, -108
lat_box_min, lat_box_max = 32, 40

# Legend labels
aerosol_labels = {
    "TOTEXTTAU": "Total AOD",
    "DUEXTTAU": "Dust",
    "BCEXTTAU": "Black carbon",
    "OCEXTTAU": "Organic carbon",
    "SSEXTTAU": "Sea salt",
    "SUEXTTAU": "Sulfate",
}

# Colors
aerosol_colors = {
    "TOTEXTTAU": "black",
    "DUEXTTAU": "#D98E04",
    "BCEXTTAU": "#7B2CBF",
    "OCEXTTAU": "#238B45",
    "SSEXTTAU": "#2B8CBE",
    "SUEXTTAU": "#D73027",
}

rho_grid = np.linspace(-1, 1, 500)

fig, ax = plt.subplots(figsize=(8, 5))

for aod_var in aod_var_list:

    corr_map = results_dict[aod_var]["spearman_corr_map"]

    # Mask for southwestern pixels
    southwest_mask = (
        (corr_map["lon"] >= lon_box_min)
        & (corr_map["lon"] <= lon_box_max)
        & (corr_map["lat"] >= lat_box_min)
        & (corr_map["lat"] <= lat_box_max)
    )

    corr_sw = corr_map.where(southwest_mask, drop=True)

    # 1D array
    correlation_values = corr_sw.compute().values.ravel()
    correlation_values = correlation_values[
        np.isfinite(correlation_values)
    ]

    if (correlation_values.size < 2
        or np.nanstd(correlation_values) == 0 ):
        
        print(f"Skipping {aod_var}: insufficient variation.")
        continue

    # PDF
    kde = stats.gaussian_kde(
        correlation_values,
        bw_method="scott"
    )

    density = kde(rho_grid)

    ax.plot(rho_grid, density, color=aerosol_colors.get(aod_var), linewidth=2.2, label=aerosol_labels.get(aod_var, aod_var) )


ax.axvline(0, color="gray", linestyle="--", linewidth=1.2, zorder=0)

ax.set_xlim(-1, 1)

ax.set_xlabel( r"Spearman correlation coefficient, $\rho$", fontsize=12)

ax.set_ylabel( "Probability density", fontsize=12)

ax.tick_params( axis="both", labelsize=11)

ax.legend(frameon=False, fontsize=10, ncol=2)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)


plt.savefig("AOD_SoilWater_Correlation_PDF_Southwest.png",dpi=300,bbox_inches="tight")

plt.show()
```

## Relative Contributions

```python
plt.close("all")


# 1. Define events


events = {
    "September 2017": ("2017-09-01", "2017-09-09"),
    "September 2020": ("2020-09-08", "2020-09-25"),
    "June 2021": ("2021-06-17", "2021-06-21"),
}

event_names = list(events.keys())


# 2. Select aerosol species


species_info = {
    "DUEXTTAU": {"title": "Dust", "label": "DuAOD / Total AOD"},
    "BCEXTTAU": {"title": "Black carbon", "label": "BcAOD / Total AOD"},
    "OCEXTTAU": {"title": "Organic carbon", "label": "OcAOD / Total AOD"},
}

species_list = list(species_info.keys())

cmap_dict = {
    "DUEXTTAU": "YlOrBr",
    "BCEXTTAU": "Reds",
    "OCEXTTAU": "Greens",
}

# 3. Plot settings


extent = [-123, -98, 25, 45]

lon_box_min, lon_box_max = -120, -108
lat_box_min, lat_box_max = 32, 40

lon_ticks = [-120, -115, -110, -105, -100]
lat_ticks = [25, 30, 35, 40, 45]

vmin = 0
n_contour_levels = 101
n_colorbar_ticks = 6


# 4. Event-mean aerosol fractions


event_contributions = {}

for event_name, (start_date, end_date) in events.items():

    ds_event = mean_aod_merra.sel(time=slice(start_date, end_date))

    if ds_event.sizes.get("time", 0) == 0:
        raise ValueError(f"No data were found for {event_name}: {start_date} to {end_date}")

    total_mean = ds_event["TOTEXTTAU"].mean(dim="time", skipna=True)
    event_contributions[event_name] = {}

    for species in species_list:
        species_mean = ds_event[species].mean(dim="time", skipna=True)
        fraction = xr.where(total_mean > 0, species_mean / total_mean, np.nan)
        fraction = fraction.where((fraction >= 0) & (fraction <= 1))
        event_contributions[event_name][species] = fraction

# 5. Maximum value for each species column


species_vmax = {}

for species in species_list:
    column_maxima = [float(event_contributions[event_name][species].max(skipna=True).values)
                     for event_name in event_names]
    species_vmax[species] = max(column_maxima)

print("Maximum value used for each column:")
for species in species_list:
    print(f"{species_info[species]['title']}: {species_vmax[species]:.4f}")


# 6. Contour levels and colorbar ticks

species_levels = {}
species_cbar_ticks = {}

for species in species_list:
    vmax = species_vmax[species]
    species_levels[species] = np.linspace(vmin, vmax, n_contour_levels)
    species_cbar_ticks[species] = np.linspace(vmin, vmax, n_colorbar_ticks)

# 7. Figure

fig, axes = plt.subplots(
    nrows=3, ncols=3, figsize=(12, 8.8),
    subplot_kw={"projection": ccrs.PlateCarree()}
)

panel_letters = ["(a)", "(b)", "(c)", "(d)", "(e)", "(f)", "(g)", "(h)", "(i)"]
panel_index = 0

# 8. Plotting maps

for row, event_name in enumerate(event_names):

    for col, species in enumerate(species_list):

        ax = axes[row, col]
        data = event_contributions[event_name][species]

        lon = data["lon"].values
        lat = data["lat"].values
        lon2d, lat2d = np.meshgrid(lon, lat)

        vmax = species_vmax[species]
        levels = species_levels[species]
        norm = Normalize(vmin=vmin, vmax=vmax)

        ax._autoscaleXon = False
        ax._autoscaleYon = False

        contour = ax.contourf(
            lon2d, lat2d, data.values,
            levels=levels, cmap=cmap_dict[species], norm=norm,
            transform=ccrs.PlateCarree(), extend="neither"
        )

        ax.set_extent(extent, crs=ccrs.PlateCarree())


        ax.add_feature(cf.COASTLINE, linewidth=0.7, zorder=3)
        ax.add_feature(cf.BORDERS, linewidth=0.7, zorder=3)
        ax.add_feature(cf.STATES, linewidth=0.5, zorder=3)

        ax.plot([lon_box_min, lon_box_max, lon_box_max, lon_box_min, lon_box_min],
            [lat_box_min, lat_box_min, lat_box_max, lat_box_max, lat_box_min],
            color="black", linewidth=1.1, transform=ccrs.PlateCarree(), zorder=4 )

        # Ticks
        ax.set_xticks(lon_ticks, crs=ccrs.PlateCarree())
        ax.set_yticks(lat_ticks, crs=ccrs.PlateCarree())
        ax.xaxis.set_major_formatter(LongitudeFormatter())
        ax.yaxis.set_major_formatter(LatitudeFormatter())
        ax.tick_params(axis="both", labelsize=9, length=3)

        # Longitude labels
        if row < 2:
            ax.tick_params(axis="x", labelbottom=False)

        # Gridlines
        ax.gridlines(xlocs=lon_ticks, ylocs=lat_ticks,
            linewidth=0.3, color="gray", alpha=0.45, linestyle="--", zorder=2 )

        # Titles with panel labels
        if row == 0:
            ax.set_title(f"{panel_letters[panel_index]}  {species_info[species]['title']}",
                         loc="left", fontsize=11, fontweight="bold", pad=6)
        else:
            ax.set_title(f"{panel_letters[panel_index]}", loc="left",
                         fontsize=11, fontweight="bold", pad=6)

        # Row labels
        if col == 0:
            ax.text(
                -0.20, 0.50, event_name,
                transform=ax.transAxes, rotation=90,
                va="center", ha="center", fontsize=12
            )

        panel_index += 1


# 9. Layout


fig.subplots_adjust(left=0.10, right=0.98, bottom=0.14, top=0.95, wspace=0.10, hspace=0.18)


# 10. Colorbars

for col, species in enumerate(species_list):

    bottom_position = axes[2, col].get_position()
    cbar_ax = fig.add_axes([bottom_position.x0, 0.060, bottom_position.width, 0.018])

    vmax = species_vmax[species]
    levels = species_levels[species]
    ticks = species_cbar_ticks[species]
    norm = Normalize(vmin=vmin, vmax=vmax)

    scalar_map = mpl.cm.ScalarMappable(norm=norm, cmap=cmap_dict[species])
    scalar_map.set_array([])

    cbar = fig.colorbar(
        scalar_map, cax=cbar_ax, orientation="horizontal",
        ticks=ticks, boundaries=levels, spacing="proportional"
    )

    cbar.set_label(species_info[species]["label"], fontsize=10)
    cbar.ax.tick_params(axis="x", labelsize=9, length=3, pad=2)
    cbar.ax.xaxis.set_major_formatter(mticker.FormatStrFormatter("%.2f"))

plt.savefig('ExtendedDataFig2_Relative_Contributions_Species_AOD.png', dpi=200, bbox_inches='tight')

plt.show()
```
